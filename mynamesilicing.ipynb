## 1. Silicing of My Name "Pathan Zaid Khan"

Using **negative indexing and slicing**.

```python
print(name[-9:11])
print(name[-8:-5])
print(name[-16:-12])
print(name[-14:6])
print(name[3:-10])
```

### Adding words and Making New Words:
3. Creating **first aid** using string concatenation.
5. Creating **thane** using string concatenation.
7. Creating **hand** using string concatenation.

```python
result1 = "first" + " " + name[-8:-5]
print(result1)
result2 = name[-14:6] + "e"
print(result2)
result3 = name[3:-10] + "d"
print(result3)
```
